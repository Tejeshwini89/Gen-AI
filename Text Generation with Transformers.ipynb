```python
# This program demonstrates how to generate text using pre-trained transformer models.

# Install required libraries
!pip install transformers torch

# Import necessary libraries
from transformers import pipeline, set_seed

# Initialize the text generation pipeline
generator = pipeline('text-generation', model='gpt2')

# Set a seed for reproducibility
set_seed(42)

# Generate text
prompt = "I deliberately want to do this everyday"
generated_text = generator(prompt, max_length=50, num_return_sequences=3)

# Print the generated text
print("Generated Text:")
for i, text in enumerate(generated_text):
    print(f"\nSequence {i+1}:")
    print(text['generated_text'])
```

```python
from transformers import pipeline, set_seed

# Initialize a more capable model (gpt2-medium) for better story coherence
# If memory is an issue, you can revert 'gpt2-medium' to 'gpt2'
generator = pipeline('text-generation', model='gpt2-medium', device=-1)

set_seed(42)

def generate_long_story(initial_prompt, target_words=1000):
    full_story = initial_prompt
    current_word_count = len(full_story.split())

    print(f"Starting story generation...")

    # We generate in chunks to maintain flow and hit the word count
    while current_word_count < target_words:
        # We provide the last 200 words as context to keep the story on track
        context = " ".join(full_story.split()[-200:])

        # Generate the next paragraph
        output = generator(
            context,
            max_new_tokens=150,
            num_return_sequences=1,
            no_repeat_ngram_size=3, # Prevents the model from getting stuck in loops
            repetition_penalty=1.2,
            do_sample=True,
            temperature=0.8, # Adds creativity
            top_k=50,
            top_p=0.95
        )

        new_text = output[0]['generated_text'][len(context):]

        # Ensure we add a paragraph break
        full_story += "\n\n" + new_text.strip()
        current_word_count = len(full_story.split())

        print(f"Progress: ~{current_word_count} words generated.")

    return full_story

# Define your moral prompt
prompt = "In a small village, there lived a boy named Leo who found a golden compass that only pointed toward the truth, even when it was painful."

story = generate_long_story(prompt, target_words=1000)

print("\n--- THE COMPASS OF TRUTH ---\n")
print(story)
```
