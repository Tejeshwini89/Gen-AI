```python
# !pip install tensorflow

import tensorflow as tf
from tensorflow.keras import layers, models

print("\n--- GAN MODEL ---")

(x_train,_),_ = tf.keras.datasets.mnist.load_data()
x_train = x_train[...,None]/255.0

G = models.Sequential([
    layers.Input((100,)),
    layers.Dense(7*7*128),
    layers.Reshape((7,7,128)),
    layers.Conv2DTranspose(64,3,2,'same',activation='relu'),
    layers.BatchNormalization(),
    layers.Conv2DTranspose(1,3,2,'same',activation='sigmoid')
])

D = models.Sequential([
    layers.Input((28,28,1)),
    layers.Conv2D(64,3,2,'same',activation='relu'),
    layers.Conv2D(128,3,2,'same',activation='relu'),
    layers.BatchNormalization(),
    layers.Flatten(),
    layers.Dense(1,activation='sigmoid')
])

D.compile("adam","binary_crossentropy",["accuracy"])
D.trainable = False

z = layers.Input((100,))
GAN = models.Model(z, D(G(z)))
GAN.compile("adam","binary_crossentropy")

print("Generator:"); G.summary()
print("\nDiscriminator:"); D.summary()
print("\nGAN:"); GAN.summary()
```
